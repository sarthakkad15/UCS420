1. Take your roll number. Extract its individual digits, and multiply each digit by 10 to form your list L

```python
roll_no = int(input('Enter your Roll Number: '))
i=roll_no
L=[]
while i!=0:
  n=i%10
  L.append(n*10)
  i=i//10
L.reverse()
```

3. Set the random seed to your own roll number using random.seed(your_roll_number) before generating your numbers — this keeps your list random-looking but reproducible and unique to you

```python
import random
random.seed(1024170117)
```

i.

```python
numbers=[random.randint(100,900) for _ in range(100)]
print(numbers)
```

ii.

```python
odd=[x for x in numbers if x%2!=0]
print(odd)
print("Length:",len(odd))
```

iii.

```python
even=[x for x in numbers if x%2==0]
print(even)
print("Length:",len(even))
```

iv.

```python
def is_prime(n):
  if n<2:
    return False
  for i in range(2,n):
    if n%i==0:
      return False
  return True

prime=[x for x in numbers if is_prime(x)]
print(prime)
print("Length:",len(prime))
```

v.


```python
freq={}
for x in numbers:
  if x in freq:
    freq[x]+=1
  else:
    freq[x]=1
most_freq=max(freq,key=freq.get)
print("Most frequent number:",most_freq)
print("Frequency:",freq[most_freq])
```
