1. Take your roll number. Extract its individual digits, and multiply each digit by 10 to form your list L

```python
roll_no = int(input('Enter your Roll Number: '))
i=roll_no
L=[]
while i!=0:
  n=i%10
  L.append(n*10)
  i=i//10
L.reverse()
```

i.


```python
print(L)
```

ii.

```python
L.append(18) # an element got added at the end
print(L)
L.insert(3,27)# an element got added at the required position
print(L)
```

iii.

```python
L.pop(2)
print(L)
L.remove(0)
print(L)
```

iv.

```python
L.sort()
print(L)
L.sort(reverse=True)
print(L)
```

v.

```python
print(L[:3])
print(L[-3:])
```

vi.

```python
average=sum(L)/len(L)
new_list=[x for x in L if x>average]
print(new_list)
```
