Q2 A. Reverse the NumPy Array

```python

import numpy as np

arr = np.array([1, 2, 3, 6, 4, 5])

reversed_arr = arr[::-1]

print("Original Array:", arr)
print("Reversed Array:", reversed_arr)
```

Q2 B. Find the most frequent value

```python

import numpy as np

x = np.array([1, 2, 3, 4, 5, 1, 2, 1, 1, 1])

values, counts = np.unique(x, return_counts=True)

max_count = counts.max()
modes = values[counts == max_count]

print("Array:", x)
print("Most frequent value(s):", modes)
print("Maximum frequency:", max_count)

for mode in modes:
    print(f"Indices of {mode}:", np.where(x == mode)[0])
```

Q4

```python

import numpy as np
Sarthak = np.linspace(10, 100, 25)

print("Array:", Sarthak)
print("Dimensions:", Sarthak.ndim)
print("Shape:", Sarthak.shape)
print("Total elements:", Sarthak.size)
print("Data type:", Sarthak.dtype)
print("Memory consumed (bytes):", Sarthak.nbytes)

print("Transpose using T:", Sarthak.T)
row_array = Sarthak.reshape(1, 25)
print("Row array shape:", row_array.shape)

transposed_array = row_array.T
print("Transposed array shape:", transposed_array.shape)
print("Transposed array:\n", transposed_array)
```

```python

import numpy as np

ucs420_Sarthak = np.array([
    [10, 20, 30, 40],
    [50, 60, 70, 80],
    [90, 15, 20, 35]
])

print("Original Array:\n", ucs420_Sarthak)

print("Mean:", np.mean(ucs420_Sarthak))
print("Median:", np.median(ucs420_Sarthak))
print("Maximum:", np.max(ucs420_Sarthak))
print("Minimum:", np.min(ucs420_Sarthak))
print("Unique elements:", np.unique(ucs420_Sarthak))

reshaped_ucs420_Sarthak = ucs420_Sarthak.reshape(4, 3)
print("\nReshaped Array (4 x 3):\n", reshaped_ucs420_Sarthak)

resized_ucs420_Sarthak = np.resize(ucs420_Sarthak, (2, 3))
print("\nResized Array (2 x 3):\n", resized_ucs420_Sarthak)
```
