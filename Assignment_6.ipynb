```python

import numpy as np

temperature = np.array([25, 28, 31, 35, 38, 27, 33, 40])

corrected_temperature = temperature + 2
print("Corrected temperatures:", corrected_temperature)

fahrenheit = (9/5) * corrected_temperature + 32
print("Temperatures in Fahrenheit:", fahrenheit)

readings_above_32 = corrected_temperature[corrected_temperature > 32]
print("Readings above 32°C:", readings_above_32)

count = np.sum(corrected_temperature > 32)
print("Number of readings above 32°C:", count)
```

Q2

```python

import numpy as np

steps = np.array([
    [5000, 6200, 7100],
    [8000, 7500, 9000],
    [4500, 5100, 4800],
    [9000, 8500, 9500]
])

print("Total steps:", np.sum(steps))

print("Mean steps:", np.mean(steps))

print("Maximum steps:", np.max(steps))
print("Minimum steps:", np.min(steps))

print("Total steps for each day:", np.sum(steps, axis=0))

print("Total steps for each user:", np.sum(steps, axis=1))

position = np.unravel_index(np.argmax(steps), steps.shape)
print("Maximum step position (user index, day index):", position)
print("Maximum step count:", steps[position])
```

```python

import numpy as np

original = np.array([1, 2, 3, 4, 5, 6])
print("(a) Original array:", original)

subset = original[1:4]
print("\n(b) Subset:", subset)

subset[0] = 999
print("\n(c) After modifying subset:")
print("Original:", original)
print("Subset:", subset)

copied = original[1:4].copy()
copied[0] = 500
print("\n(d) After modifying copied array:")
print("Original:", original)
print("Copied:", copied)

matrix = np.arange(1, 13).reshape(3, 4)
print("\n(e) 3 x 4 Matrix:\n", matrix)

print("\n(f) First row:", matrix[0, :])
print("Last row:", matrix[-1, :])
print("Second column:", matrix[:, 1])
print("Rows 1-2, columns 2-3:\n", matrix[0:2, 1:3])

flat_copy = matrix.flatten()
flat_view = matrix.ravel()
print("\n(g) Using flatten():", flat_copy)
print("Using ravel():", flat_view)

flat_view[0] = 100
print("\n(h) After modifying ravel result:")
print("Original matrix:\n", matrix)
print("Ravel result:", flat_view)

flat_copy[1] = 200
print("\n(i) After modifying flatten result:")
print("Original matrix:\n", matrix)
print("Flatten result:", flat_copy)

print("\n(j) Shape:", matrix.shape)
print("Dimensions:", matrix.ndim)
print("Total elements:", matrix.size)
print("Data type:", matrix.dtype)
```

```python

import numpy as np

X = np.array([
    [6, 70, 3],
    [5, 50, 6],
    [8, 80, 2],
    [4, 30, 8]
])

y = np.array([40, 65, 30, 85])

print("(a) Shape of X:", X.shape)
print("Dimensions of X:", X.ndim)

XT = X.T
print("\n(b) Transpose of X:\n", XT)

product = XT @ X
print("\n(c) X.T @ X:\n", product)

inverse = np.linalg.inv(product)
print("\n(d) Inverse of X.T @ X:\n", inverse)

beta = inverse @ XT @ y
print("\n(e) Regression coefficients:", beta)

print("\n(f) Coefficients:")
print("Sleep coefficient:", beta[0])
print("Activity coefficient:", beta[1])
print("Stress coefficient:", beta[2])

new_user = np.array([5, 40, 7])
predicted_score = new_user @ beta
print("\n(g) Predicted assistance score:", predicted_score)
```
