1. Build a simple Cognitive FAQ System using Pandas DataFrame.

```python
import pandas as pd
```

```python
roll_number = input("Enter your roll number: ")

print("Roll Number:", roll_number)
print("Last two digits:", roll_number[-2:])
```

```python
categories = ["billing", "account", "general"]

last_two_digits = [int(d) for d in roll_number[-2:]]

personalized_categories = [
    categories[d % 3] for d in last_two_digits
]

print("Last two digits:", last_two_digits)
print("Assigned categories:", personalized_categories)
```

```python
fixed_entries = [
    {
        "question": "what is the annual fee",
        "answer": "The annual fee is Rs 500.",
        "keywords": "fee cost price charge",
        "category": "billing"
    },
    {
        "question": "how to reset password",
        "answer": "Go to Settings > Reset Password.",
        "keywords": "password reset login",
        "category": "account"
    },
    {
        "question": "what are your working hours",
        "answer": "We are open 9 AM to 5 PM.",
        "keywords": "hours timing open time",
        "category": "general"
    },
    {
        "question": "how can i pay the fee",
        "answer": "You can pay via UPI, card, or net banking.",
        "keywords": "pay payment upi fee",
        "category": "billing"
    }
]
```

```python
personalized_entries = []

for digit, category in zip(last_two_digits, personalized_categories):

    if category == "billing":
        question = "how can i check my recent payment"
        answer = "You can check your recent payment details from the billing section."
        keywords = "payment billing transaction"

    elif category == "account":
        question = "how do i update my registered mobile number"
        answer = "Go to Account Settings and update your registered mobile number."
        keywords = "mobile number update"

    else:
        question = "how can i contact customer support"
        answer = "You can contact customer support through the Help and Support section."
        keywords = "support help contact"

    personalized_entries.append({
        "question": question,
        "answer": answer,
        "keywords": keywords,
        "category": category
    })

personalized_entries
```

```python
all_entries = fixed_entries + personalized_entries

df = pd.DataFrame(all_entries)

print("Final FAQ DataFrame:")
display(df)
```

 2. Implement a scoring function that takes a query string and returns all matching entries ranked by confidence

```python
def score_query(query, df):
    query_words = set(query.lower().split())

    results = []

    for index, row in df.iterrows():

        keywords = set(row["keywords"].lower().split())
        question_words = set(row["question"].lower().split())

        searchable_words = keywords.union(question_words)

        matched_words = query_words.intersection(searchable_words)

        score = len(matched_words) / len(query_words)

        if score > 0:
            results.append({
                "index": index,
                "question": row["question"],
                "answer": row["answer"],
                "category": row["category"],
                "confidence": score
            })

    results_df = pd.DataFrame(results)

    if not results_df.empty:
        results_df = results_df.sort_values(
            by="confidence",
            ascending=False
        ).reset_index(drop=True)

    return results_df
```

```python
query = input("Enter your FAQ query: ")

results = score_query(query, df)

if results.empty:
    print("No matching FAQ found.")
else:
    print("\nMatching FAQs ranked by confidence:")
    display(results)
```

6. Modify your Q2 scoring function so that if two or more entries tie for the highest score, it does not silently pick one — it prints all matching entries instead, so the user can see every equally good match

```python
def score_query_with_tie(query, df):

    query_words = set(query.lower().split())

    results = []

    for index, row in df.iterrows():

        keywords = set(row["keywords"].lower().split())
        question_words = set(row["question"].lower().split())

        searchable_words = keywords.union(question_words)

        matched_words = query_words.intersection(searchable_words)

        score = len(matched_words) / len(query_words)

        if score > 0:
            results.append({
                "index": index,
                "question": row["question"],
                "answer": row["answer"],
                "category": row["category"],
                "confidence": score
            })

    results_df = pd.DataFrame(results)

    if results_df.empty:
        print("No matching FAQ found.")
        return results_df

    results_df = results_df.sort_values(
        by="confidence",
        ascending=False
    ).reset_index(drop=True)

    highest_score = results_df["confidence"].max()

    top_matches = results_df[
        results_df["confidence"] == highest_score
    ]

    print(f"Highest confidence score: {highest_score:.2f}")

    if len(top_matches) > 1:
        print("\nTIE DETECTED!")
        print("Multiple FAQs have the same highest confidence:")
    else:
        print("\nUnique highest-confidence match:")

    display(top_matches)

    return results_df
```

```python
print(" TIE EXAMPLE ")

tie_results = score_query_with_tie("fee", df)
```

```python
print("NON-TIE EXAMPLE")

non_tie_results = score_query_with_tie(
    "password reset login",
    df
)
```
